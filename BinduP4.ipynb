```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Student":["A","B","C","D","E"],
    "Gender":["male","female","male","female","male"],
    "StudyHours": [5,6,np.nan,4,7],
    "Attendance":[80,85,75,np.nan,90],
    "Marks":[75,82,68,55,np.nan],
    "Result":["Pass","Pass","Pass","Fail","Pass"]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
print("\n Missing values: ")
print(df.isnull().sum())
```

```python
df["StudyHours"] = df["StudyHours"].fillna(
    df["StudyHours"].mean()                  #mean
)

print("Dataset After Mean Imputation:")
print(df)
```

```python
df["Marks"] = df["Marks"].fillna(
    df["Marks"].median()          #Median
)

print("Dataset After Median Imputation:")
print(df)
```

```python
df["Attendance"] = df["Attendance"].ffill()  #Farward Fill
print("Dataset After Farward Fill:")    
print(df)
```

```python
#Encode Gender
df["Gender"] = df["Gender"].map({
    "male":1,
    "female":0
})

print("Dataset After Gender Encoding:")
print(df)
```

```python
df["Result"] = df["Result"].map({"Pass":1,"Fail":0})
print("Dataset After Result Encoding")
print(df)
```

```python
print("Missing Values after preprocessing")
print(df.isnull().sum())
```

```python
print("Final preprocessed dataset")
print(df)
```
